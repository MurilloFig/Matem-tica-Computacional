### Multiplicação de Matrizes

$$ M = \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix} \cdot \begin{bmatrix} 1 & 1 \\ 0 & 1 \end{bmatrix} $$

#### Utilizando o Numpy para fazer essa multiplicação

```python
import numpy as np

matriz2 = np.array([[1,0],[1,1]])
matriz1 = np.array([[0,1],[-1,0]])

matriz = matriz1 @ matriz2
print(matriz)
```

Com isso temos a multiplicação dessas duas matrizes sendo igual a: 
\begin{bmatrix} 
1 & 1 \\ 
-1 & 0 
\end{bmatrix} 

Com isso temos a multiplicação da matriz de trasformação da rotação e em seguida o cisalhamento

#### Agora vamos inverter a ordem da transformação

```python
import matplotlib.pyplot as plt

plt.quiver(0,0,0,1,scale=1,scale_units='xy',angles='xy')
plt.quiver(0,0,1,0,scale=1,scale_units='xy',angles='xy')
plt.xlim(-3,3)
plt.ylim(-3,3)
plt.grid(True)
plt.axhline(0,color='black',linewidth=1)
plt.axvline(0,color='black',linewidth=1)
plt.text(-0.20,0.40,'j',fontweight='bold')
plt.text(0.40,-0.30,'i',fontweight='bold')
plt.gca().set_aspect('equal', adjustable='box')
```

### Vetor i:

A seguir é feito a concatenação dos dois vetores e então é criada uma matriz transformação:

\begin{bmatrix}
1 & 1 \\
-1 & 0
\end{bmatrix}

### Agora fazemos a rotação após o cisalhamento

Para isso sabemos que para se criar a matriz transformação da rotação foi feito a multiplicação da matriz de Rotação pela matriz de Cisalhamento e como a multiplicação de Matrizes não é associativa basta fazermo a multiplicação da matriz de Cisalhamento pela de Rotação agora e teremos sua matriz de transformação para multiplicarmos pela matriz inicial 

$$ M = \begin{bmatrix} 1 & 1 \\ 0 & 1 \end{bmatrix} \cdot \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix} $$

```python
import numpy as np

matriz1 = np.array([[1,1],[0,1]])
matriz2 = np.array([[0,1],[-1,0]])

matriz = matriz1 @ matriz2
print(matriz)
```

Logo temos a seguinte matriz de transformação:
\begin{bmatrix}
-1 & 1 \\
-1 & 0
\end{bmatrix}

### Calculando o produto escalar entre os dois vetores

#### Vamos utilizar um vetor arbitrário

Vetor escolhido:
\begin{bmatrix}
2 \\
3
\end{bmatrix}

#### Vamos fazer a transformação desse vetor para uma rotação de 90° no eixo x

$$ M = \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix} \cdot \begin{bmatrix} 2 \\ 3 \end{bmatrix} $$

```python
import numpy as np

matriz_transformação = np.array([[0,1],
                                 [-1,0]])
vetor = np.array([2,3])

vetor_resultante = matriz_transformação @ vetor
print(vetor_resultante)
```

Com isso obtemos o seguinte vetor

\begin{bmatrix}
3 \\
-2
\end{bmatrix}

#### Com isso nós que fazer então a análise do produto escalar dos dois vetores

```python
import numpy as np

v1 = np.array([2, 3])
v2 = np.array([3, -2])

dot = np.dot(v1, v2)
print(dot)
```

#### Resultado do Produto escalar:

$\vec {v1} \cdot \vec {v2} = 0$

### Conclusão

Podemos ver que o resultado do produto escalar é 0, Portanto fica claro que esses dois vetores são perpendiculares o que de fato faz sentido por termos rotacionado um deles em 90°
