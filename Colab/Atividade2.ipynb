Primeiro vamos começar fazendo a automatização das operações matemáticas envolvidas para termos os dados necessários para plotar o gráfico

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

altura = [1.4,1.9,3.2]
slope = 0.64
intercept = 0.1
weight = [0.5,2.3,2.9]
Learning_Rate = 0.01
gradients_list = []
step_history = []
squared_errors_list = []
intercept_history = []
total_error_history = []
gradient_sum_history = []
max_steps = 50

def compute_gradients(altura, intercept, slope, weight):
    gradients_list = []
    squared_errors_list = []
    for i in range(0,3):
        squa = (altura[i] - (intercept + (slope * weight[i]))) ** 2
        derivada = -2 * (altura[i] - (intercept + (slope * weight[i])))
        gradients_list.append(derivada)
        squared_errors_list.append(squa)
    return sum(squared_errors_list), sum(gradients_list), squared_errors_list, gradients_list

def gradient_step(altura, slope, weight, Learning_Rate,
                  intercept, intercept_history, step_history):
    intercept_history.append(intercept)
    sum_squa,deri_sum,squared_errors_list, gradients_list = compute_gradients(altura, intercept, slope, weight)

    step_size = deri_sum * Learning_Rate
    step_history.append(step_size)

    intercept = intercept - step_size

    return intercept, sum_squa, deri_sum, step_size,squared_errors_list,gradients_list


for i in range(max_steps):
    intercept, sum_squa, deri_sum, step_size, squared_errors_list, gradients_list = gradient_step(altura, slope, weight, Learning_Rate, intercept,intercept_history, step_history)

    total_error_history.append(sum_squa)
    gradient_sum_history.append(deri_sum)
    print("------------------------------------------------")
    print("Valores das derivadas:", np.around(gradients_list, 2))
    print("Soma dos quadrados residuais:", np.around(sum_squa, 2))
    print("Soma das derivadas:", np.around(deri_sum, 2))
    print("Step size:", np.around(step_size, 2))
    print("Intercepts até agora:", np.around(intercept_history, 2))
    print("Novo intercept:", np.around(intercept, 2))
    print("------------------------------------------------\n")
    print(gradients_list)
intercept_history.append(intercept)
squared_errors_list.append(compute_gradients(altura,intercept,slope,weight)[0])
gradient_sum_history.append(compute_gradients(altura,intercept,slope,weight)[1])
````

#### Letra B.1)

Este código implementa o Gradiente Descendente para regressão linear, atualizando dois parâmetros ao mesmo tempo. O objetivo é encontrar os valores de intercept e slope que minimizam a Soma dos Quadrados Residuais.

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

altura = [1.4,1.9,3.2]
slope = 1.0
intercept = 0.0
weight = [0.5,2.3,2.9]
Learning_Rate = 0.01
gradients_list = []
step_intercept_history = []
step_slope_history = []
squared_errors_list = []
intercept_history = []
slope_history = []
total_error_history = []
gradient_sum_history = []
max_steps = 500

def print_gradient_details(altura, weight, intercept, slope, Learning_Rate):

    soma_residuos = 0.0
    soma_intercept = 0.0
    soma_slope = 0.0

    for i in range(len(altura)):
        y = altura[i]
        x = weight[i]
        prediction = intercept + slope * x
        residual = y - prediction

        soma_residuos = soma_residuos + residual**2
        soma_intercept = soma_intercept + residual
        soma_slope = (soma_slope + x * residual)

    # Derivadas
    deriv_intercept = -2 * soma_intercept
    deriv_slope = -2 * soma_slope

    # Step size
    step_i = deriv_intercept * Learning_Rate
    step_s = deriv_slope * Learning_Rate

    # Atualizando intercept e slope
    new_intercept = intercept - step_i
    new_slope = slope - step_s

    print("--------------------------------------------------")
    print("Intercept atual: ",intercept)
    print("Slope atual: " ,slope)
    print("Soma dos quadrados residuais: ",round(soma_residuos, 6))
    print("Derivada do Intercept: "  ,round(deriv_intercept, 6))
    print("Derivada do Slope:     " ,round(deriv_slope, 6))
    print("Step Size Intercept = "  ,round(step_i, 6))
    print("Step Size Slope     = "  ,round(step_s, 6))
    print("Novo Intercept:       ", round(new_intercept, 6))
    print("Novo Slope:           ", round(new_slope, 6))
    print("--------------------------------------------------\n")

    return new_intercept, new_slope, deriv_intercept, deriv_slope


for i in range(max_steps):
    new_intercept, new_slope, deriv_intercept, deriv_slope = print_gradient_details(altura, weight, intercept, slope, Learning_Rate)

    intercept = new_intercept
    slope = new_slope
````

Bom temos como base que a soma dos quadrados residuais é igual a:

$Sum of Squared Residual = (observed - predicted)^2$

Sabemos que o observed é igual a altura e o preditcted é :

$Predicted = Intercept + slope \cdot weight$

O que está presenta na seguinte linha de código:
```python
prediction = intercept + slope * x

E com isso nós podemos calcular a soma dos residuos, assim:

$SSR(\text{intercept}, \text{slope})
= \sum_{i=1}^{n} \left( y_i - \left( \text{intercept} + \text{slope} \cdot x_i \right) \right)^2$

$Sum\,\, of \,\,Squared \,\,Residuals(intercept) = (1.4 - (intercept + slope \cdot 0.5))^2 + (1.9 - (intercept + slope \cdot 2.3))^2 + (3.2 - (intercept + slope \cdot 2.9))^2 $

$Sum \,\, of \,\,Squared \,\,Residuals(slope) =  (1.4 - (intercept + slope \cdot 0.5))^2 +  (1.9 - (intercept + slope \cdot 2.3))^2 +  (3.2 - (intercept + slope \cdot 2.9))^2 $

$Sum of Squared Residual = (altura - (Intercept + slope \cdot weight))^2$

Sendo representada pelas seguintes linhas de código:
```python
residual = y - prediction
soma_residuos = soma_residuos + residual**2
soma_intercept = soma_intercept + residual
soma_slope = (soma_slope + x * residual)

Além disso após esses resultados, iremos fazer o cálculo das derivadas do slope e do intercept:

$\frac{\partial SSR}{\partial \text{slope}}
= -2 \sum_{i=1}^{n}
x_i \left( y_i - \left( \text{intercept} + \text{slope} \cdot x_i \right) \right)$


$\frac{d}{d\,\,slope} Sum\,\, of \,\,Squared \,\,Residuals = \frac{d}{d\,\,slope} (1.4 - (intercept + slope \cdot 0.5))^2 + \frac{d}{d\,\,slope} (1.9 - (intercept + slope \cdot 2.3))^2 + \frac{d}{d\,\,slope} (3.2 - (intercept + slope \cdot 2.9))^2 $

$\frac{\partial SSR}{\partial \text{intercept}}
= -2 \sum_{i=1}^{n}
\left( y_i - \left( \text{intercept} + \text{slope} \cdot x_i \right) \right)$


$\frac{d}{d\,\,intercept} Sum\,\, of \,\,Squared \,\,Residuals = \frac{d}{d\,\,intercept} (1.4 - (intercept + slope \cdot 0.5))^2 + \frac{d}{d\,\,intercept} (1.9 - (intercept + slope \cdot 2.3))^2 + \frac{d}{d\,\,intercept} (3.2 - (intercept + slope \cdot 2.9))^2 $

#### Letra B.2)

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML


lista_alturas = [1.4, 1.9, 3.2]
lista_weight   = [0.5, 2.3, 2.9]
intercept_inicial = 0.0
slope_inicial = 1.0
learning_rate = 0.01
max_steps = 500

def step_gradient_descendente(lista_alturas,lista_weight,intercept_atual,slope_atual,learning_rate):


    soma_dos_quadrados_residuais = 0.0
    soma_da_derivada_intercept = 0.0
    soma_da_derivada_slope = 0.0

    for indice in range(len(lista_alturas)):
        y_real = lista_alturas[indice]
        x_real = lista_weight[indice]

        # previsão da reta
        y_previsto = intercept_atual + slope_atual * x_real

        # resíduo = taxa de erro da reta no ponto atual
        residuo = y_real - y_previsto

        # acumula o erro ao quadrado
        soma_dos_quadrados_residuais = soma_dos_quadrados_residuais + residuo**2

        # partes que aparecem nas derivadas
        soma_da_derivada_intercept = soma_da_derivada_intercept + residuo
        soma_da_derivada_slope = soma_da_derivada_slope + x_real * residuo

    # fórmulas da derivada da soma dos quadrados dos resíduos
    derivada_intercepto = -2 * soma_da_derivada_intercept
    derivada_inclinacao = -2 * soma_da_derivada_slope

    # tamanho do step
    step_size_intercept = derivada_intercepto * learning_rate
    step_size_slope = derivada_inclinacao * learning_rate

    # atualizando intercept e o slope
    novo_intercept = intercept_atual - step_size_intercept
    novo_slope = slope_atual - step_size_slope

    return novo_intercept, novo_slope, soma_dos_quadrados_residuais

lista_intercepts = [intercept_inicial]
lista_slopes = [slope_inicial]
lista_erros  = []

intercept_atual = intercept_inicial
slope_atual = slope_inicial

for passo in range(max_steps):
    novo_intercept, novo_slope, erro = step_gradient_descendente(lista_alturas,lista_weight,intercept_atual,slope_atual,learning_rate)

    # guarda no histórico
    lista_intercepts.append(novo_intercept)
    lista_slopes.append(novo_slope)
    lista_erros.append(erro)

    # atualiza para o próximo step
    intercept_atual = novo_intercept
    slope_atual = novo_slope

figura, eixo = plt.subplots()
eixo.grid(True)

# pontos
eixo.scatter(lista_weight, lista_alturas, s=70)
eixo.set_xlabel("Weight")
eixo.set_ylabel("Height")


x_para_reta = np.linspace(min(lista_weight) - 0.5, max(lista_weight) + 0.5, 100)

y_para_reta = lista_intercepts[0] + lista_slopes[0] * x_para_reta
linha_regressao, = eixo.plot(x_para_reta, y_para_reta, linewidth=2)

def atualizar_frame(indice_frame):
    intercept = lista_intercepts[indice_frame]
    slope = lista_slopes[indice_frame]

    y_para_reta = intercept + slope * x_para_reta
    linha_regressao.set_ydata(y_para_reta)

    titulo = "Passo: " + str(indice_frame)
    titulo = titulo + " | intercept: " + str(round(intercept, 3))
    titulo = titulo + " | slope: " + str(round(slope, 3))
    eixo.set_title(titulo)

    return linha_regressao,


animacao = FuncAnimation(figura,atualizar_frame,frames=len(lista_intercepts),interval=75,blit=True)

plt.show()

HTML(animacao.to_jshtml())

````

### Gradiente Stochatico e Mini Batch

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

lista_alturas = [1.4, 1.9, 3.2]
lista_weight  = [0.5, 2.3, 2.9]

numero_pontos = len(lista_alturas)

intercept = 0.0
slope = 1.0
taxa_aprendizado = 0.01
max_passos = 50

np.random.seed(0)


def um_step_gradiente_descendente(indice_cada_ponto,lista_alturas,lista_weight,intercept_atual,slope_atual,taxa_aprendizado):

    soma_dos_quadrados_residuais = 0.0
    soma_da_derivada_intercept = 0.0
    soma_da_derivada_slope = 0.0

    for indice in indice_cada_ponto:
        altura = lista_alturas[indice]
        weight = lista_weight[indice]

        y_previsto = intercept_atual + slope_atual * weight
        residuo = altura - y_previsto

        soma_dos_quadrados_residuais = soma_dos_quadrados_residuais + residuo**2
        soma_da_derivada_intercept = soma_da_derivada_intercept + residuo
        soma_da_derivada_slope = soma_da_derivada_slope + weight * residuo

    deri_intercept = -2 * soma_da_derivada_intercept
    deri_slope = -2 * soma_da_derivada_slope

    step_intercept = deri_intercept * taxa_aprendizado
    step_slope = deri_slope * taxa_aprendizado

    novo_intercept = intercept_atual - step_intercept
    novo_slope = slope_atual - step_slope

    return novo_intercept, novo_slope, soma_dos_quadrados_residuais


lista_intercept_sgd = [intercept]
lista_slope_sgd = [slope]

intercept_atual = intercept
slope_atual = slope

#sgd
for passo in range(max_passos):
    chute = np.random.randint(0, numero_pontos)
    chutes_plural = [chute]

    novo_intercept, novo_slope, erro = um_step_gradiente_descendente(chutes_plural,lista_alturas,lista_weight,intercept_atual,slope_atual,taxa_aprendizado)
    lista_intercept_sgd.append(novo_intercept)
    lista_slope_sgd.append(novo_slope)

    intercept_atual = novo_intercept
    slope_atual = novo_slope


lista_intercept_mb = [intercept]
lista_slope_mb = [slope]

intercept_atual = intercept
slope_atual = slope

#mini-batch
for passo in range(max_passos):
    chutes_plural = np.random.choice(numero_pontos, size=2, replace=False)

    novo_intercept, novo_slope, erro = um_step_gradiente_descendente(chutes_plural,lista_alturas,lista_weight,intercept_atual,slope_atual,taxa_aprendizado)
    lista_intercept_mb.append(novo_intercept)
    lista_slope_mb.append(novo_slope)

    intercept_atual = novo_intercept
    slope_atual = novo_slope


def criar_animacao(nome_metodo, lista_intercepts, lista_slopes):

    figura, eixo = plt.subplots()

    eixo.scatter(lista_weight, lista_alturas, s=80)
    eixo.set_xlabel("Weight")
    eixo.set_ylabel("Height")
    eixo.grid(True)


    x_para_reta = np.linspace(min(lista_weight) - 0.5,max(lista_weight) + 0.5, 100)


    y_para_reta = lista_intercepts[0] + lista_slopes[0] * x_para_reta
    linha_regressao, = eixo.plot(x_para_reta, y_para_reta, linewidth=2)

    def atualizar_frame(indice_frame):
        intercepto = lista_intercepts[indice_frame]
        inclinacao = lista_slopes[indice_frame]

        y_para_reta = intercepto + inclinacao * x_para_reta
        linha_regressao.set_ydata(y_para_reta)

        titulo = nome_metodo + " | passo: " + str(indice_frame)
        titulo = titulo + " | intercept: " + str(round(intercepto, 3))
        titulo = titulo + " | slope: " + str(round(inclinacao, 3))
        eixo.set_title(titulo)

        return linha_regressao,

    animacao = FuncAnimation(figura,atualizar_frame,frames=len(lista_intercepts),interval=100,blit=True)
    plt.close()
    return animacao

animacao_sgd = criar_animacao(
    "SGD (1 sample)",
    lista_intercept_sgd,
    lista_slope_sgd
)

display(HTML(animacao_sgd.to_jshtml()))

animacao_mb = criar_animacao(
    "Mini-batch (2 samples)",
    lista_intercept_mb,
    lista_slope_mb
)

display(HTML(animacao_mb.to_jshtml()))
````

Bom as expliacações a cerca da derivada e dos residuos praticamente igual a do gradiente descendente anteriormente visto, porém agora algumas situações vão mudar primeiro teremos fixos inicialmente o intercept = 0.0 e o slope = 1.0

Além disso, no caso do SGD ou Stochastic Gradient Descendent teremos que pegar um ponto aleatório a cada passo como mostrado na seguinte linha de código:
```python
chute = np.random.randint(0, numero_pontos)

````python
import math
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML


dosages = [0.0, 0.5, 1.0]
targets = [0.0, 1.0, 0.0]
w1 = 3.34
b1 = -1.43
w2 = -3.53
b2 = 0.57
w31 = -1.22
w32 = -2.30
b3 = 0.0

taxa_aprendizado = 0.1
max_iteracoes = 200
limiar_step = 0.0001

def softplus(x):
    return math.log(1.0 + math.exp(x))

def rede(x, b3_atual):
    # neurônio 1
    z1 = w1 * x + b1
    h1 = softplus(z1)

    # neurônio 2
    z2 = w2 * x + b2
    h2 = softplus(z2)

    # saída
    y = w31 * h1 + w32 * h2 + b3_atual
    return y

def calcula_gradiente_b3(bias_atual):
    soma_grad = 0.0
    soma_erro2 = 0.0

    for i in range(len(dosages)):
        x = dosages[i]
        t = targets[i]

        y = rede(x, bias_atual)
        erro = t - y

        soma_erro2 = soma_erro2 + erro * erro
        soma_grad = soma_grad + (-2.0 * erro)

    return soma_grad, soma_erro2


hist_b3 = [b3]   # não me esquece murillo pra animação
hist_loss = []

for epoca in range(max_iteracoes):
    grad_b3, perda = calcula_gradiente_b3(b3)

    step = taxa_aprendizado * grad_b3
    novo_b3 = b3 - step

    print("-----------------------------")
    print("Iteracao:", epoca)
    print("b3 antigo:", round(b3, 4))
    print("gradiente b3:", round(grad_b3, 4))
    print("step size:", round(step, 4))
    print("b3 novo:", round(novo_b3, 4))
    print("loss:", round(perda, 4))
    print("-----------------------------\n")

    hist_b3.append(novo_b3)
    hist_loss.append(perda)
    b3 = novo_b3

    if abs(step) < limiar_step:
        break


x_plot = np.linspace(0.0, 1.0, 200)

fig, ax = plt.subplots()
ax.set_xlabel("Dosage")
ax.set_ylabel("Output")
ax.set_title("Neural Network (otimizando b3)")
ax.grid(True)

# pontos alvo
ax.scatter(dosages, targets, color="cyan", edgecolors="k", zorder=3)

# curva inicial
y_plot = []
for x in x_plot:
    y_plot.append(rede(x, hist_b3[0]))
linha, = ax.plot(x_plot, y_plot, color="green", linewidth=3)

ax.set_ylim(-2.5, 1.2)

def atualiza(frame):
    b3_frame = hist_b3[frame]

    y_plot = []
    for x in x_plot:
        y_plot.append(rede(x, b3_frame))

    linha.set_ydata(y_plot)

    titulo = "Iteracao: " + str(frame) + " | b3 = " + str(round(b3_frame, 3))
    ax.set_title(titulo)
    plt.close()
    return linha,

anim = FuncAnimation(fig, atualiza,frames=len(hist_b3),interval=200,blit=True)

plt.show()
HTML(anim.to_jshtml())
````

Logo após a definicção dos b e w, nós definimos nossa função de ativação que é similar a ReLU sendo ela a softplus, ela basicamente ao receber entradas negativas as joga para próximo de zero ao receber entredas positivas retorna valores maiores ou quase iguais a entrada inicial, sendo ela representada pela seguinte função:
```python
def softplus(x):

Após isso nós implementamos a rede:
```python
def rede(x, b3_atual):

    z2 = w2 * x + b2
    h2 = softplus(z2)

```
Aqui podemos ver nossos dois neuronios e a saída da rede, onde:
```python
z1 = w1 * x + b1
h1 = softplus(z1)
```
É o neurônio 1 onde sabemos que w1 é o peso que vai ser exercido sobre a entrada x e b1 é o bias para o ajuste da saída desse neurônio, e :
```python
z2 = w2 * x + b2
h2 = softplus(z2)
```
É o neurônio 2 onde sabemos que w2 é o peso que vai ser exercido sobre a entrada x e b2 é o bias para o ajuste da saída desse neurônio, e:
```python
y = w31 * h1 + w32 * h2 + b3_atual
```
É a saída dessa rede onde não sabemos o bias de saída e iremos descobri-lo

Após isso vamos fazer o calcúlo do perda e do gradiente de b3
```python
def calcula_gradiente_b3(bias_atual):
    soma_grad = 0.0
    soma_erro2 = 0.0

    for i in range(len(dosages)):
        x = dosages[i]
        t = targets[i]

        y = rede(x, bias_atual)
        erro = t - y
        
        soma_erro2 = soma_erro2 + erro * erro
        soma_grad = soma_grad + (-2.0 * erro)

    return soma_grad, soma_erro2
```
Para o que isso serve ? Isso serve para descobrir o quanto a rede está "errada", essa função em forma matemática é escrita da seguinte forma:

$SSR(b_3) = \sum_{i=1}^{3} \bigl(t_i - y(x_i; b_3)\bigr)^2$

onde $t_i - y(x_i; b_3)$ é representado por
```python
erro = t - y
```
E $SSR(b_3) = \sum_{i=1}^{3} \bigl(t_i - y(x_i; b_3)\bigr)^2$ é representado por :
```python
soma_erro2 = soma_erro2 + erro * erro
```

Sendo esse o calculo da soma dos residuos dessa rede e como feito no gradiente descendente temos que calcular também a derivada dessa soma representada por :

$\frac{d\,SSR}{db_3}
= -2 \sum_{i=1}^{3} \bigl(t_i - y(x_i; b_3)\bigr)$

E ao representa-la no código teremos:
```python
soma_grad = soma_grad + (-2.0 * erro)
```

Após isso tudo podemos ver que o backpropagation não é nada demais afinal:

$y = (\text{tudo somado}) + b_3
\qquad \Rightarrow \qquad
\frac{\partial y}{\partial b_3} = 1$

Portanto qualquer alteração em b3 altera diretamente a saída dessa rede

Após esses cálculos nós iremos começar o treinamento:
```python
for epoca in range(max_iteracoes):

        break
```

Bom após usarmos a função que criamos anteriormente nós calculamos o step usando a seguinte equação:

$\text{step} = \eta \cdot \frac{d\,SSR}{db_3}$

Representada na seguinte linha:
```python
step = taxa_aprendizado * grad_b3
```

Agora vamos descobrir o novo b3 sendo utilizada a seguinte equação:

$b_3^{\text{novo}} = b_3^{\text{antigo}} - \eta \cdot \frac{d\,SSR}{db_3}$

Sendo representada por :
```python
novo_b3 = b3 - step
```

Por fim nós temos um critério de parada onde se o gradiene for muito pequeno chegamos no minimo da função de perda então paramos
```python
if abs(step) < limiar_step:
    break
```

````python
import math
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

dosagens = [0.0, 0.5, 1.0]
alvos    = [0.0, 1.0, 0.0]
w1 = 3.34
b1 = -1.43
w2 = -3.53
b2 = 0.57
learning_rate = 0.1
max_iteracoes = 500
tolerancia = 0.0001

# chute inicial simples
b3 = -1.0
w3 = -1.0
w4 = -1.0
hist_b3 = [b3]
hist_w3 = [w3]
hist_w4 = [w4]
hist_erro = []

def softplus(x):
    return math.log(1.0 + math.exp(x))

def rede(dose, b3, w3, w4):
    h1 = softplus(w1 * dose + b1)
    h2 = softplus(w2 * dose + b2)
    saida = w3 * h1 + w4 * h2 + b3
    return saida, h1, h2

def calcula_gradientes(dosagens, alvos, b3, w3, w4):
    erro_total = 0.0
    grad_b3 = 0.0
    grad_w3 = 0.0
    grad_w4 = 0.0

    for x, y_real in zip(dosagens, alvos):
        y_pred, h1, h2 = rede(x, b3, w3, w4)
        erro = y_real - y_pred
        erro_total = erro_total + erro**2

        # derivadas da soma dos quadrados dos resíduos
        grad_b3 = grad_b3 + (-2.0 * erro * 1.0)
        grad_w3 = grad_w3 + (-2.0 * erro * h1)
        grad_w4 = grad_w4 + (-2.0 * erro * h2)

    return erro_total, grad_b3, grad_w3, grad_w4


for passo in range(max_iteracoes):
    erro_total, grad_b3, grad_w3, grad_w4 = calcula_gradientes(dosagens, alvos, b3, w3, w4)

    passo_b3 = learning_rate * grad_b3
    passo_w3 = learning_rate * grad_w3
    passo_w4 = learning_rate * grad_w4

    print("Iteracao:", passo + 1)
    print("Erro total:", round(erro_total, 6))
    print("antigo b3:", b3)
    print("antigo w3:", w3)
    print("antigo w4:", w4)
    print("Step Size b3:", passo_b3)
    print("Step Size w3:", passo_w3)
    print("Step Size w4:", passo_w4)

    b3 = b3 - passo_b3
    w3 = w3 - passo_w3
    w4 = w4 - passo_w4

    print("New b3:", b3)
    print("New w3:", w3)
    print("New w4:", w4)
    print("-----------------------------------")

    hist_b3.append(b3)
    hist_w3.append(w3)
    hist_w4.append(w4)
    hist_erro.append(erro_total)

    if max(abs(passo_b3), abs(passo_w3), abs(passo_w4)) < tolerancia:
        print("Parou pela condicao de parada no passo", passo + 1)
        break

x_grid = np.linspace(0.0, 1.0, 200)

fig, ax = plt.subplots()
ax.set_title("Neural Network - Gradiente Descendente")
ax.set_xlabel("Dosage")
ax.set_ylabel("Output")
ax.set_xlim(-0.05, 1.05)
ax.set_ylim(-0.2, 1.2)
ax.grid(True)

# pontos alvo
ax.scatter(dosagens, alvos, color="blue", s=60)

# curva inicial
y_ini = []
for x in x_grid:
    y_pred, _, _ = rede(x, hist_b3[0], hist_w3[0], hist_w4[0])
    y_ini.append(y_pred)
linha, = ax.plot(x_grid, y_ini, linewidth=2, color="green")

def atualiza(frame):
    b3 = hist_b3[frame]
    w3 = hist_w3[frame]
    w4 = hist_w4[frame]

    y_vals = []
    for x in x_grid:
        y_pred, _, _ = rede(x, b3, w3, w4)
        y_vals.append(y_pred)

    linha.set_ydata(y_vals)

    titulo = "Passo: " + str(frame) \
             + " | b3: " + str(round(b3, 3)) \
             + " | w3: " + str(round(w3, 3)) \
             + " | w4: " + str(round(w4, 3))
    ax.set_title(titulo)
    return linha,

animacao = FuncAnimation(fig, atualiza,frames=len(hist_b3),interval=120,blit=True)

HTML(animacao.to_jshtml())
````
